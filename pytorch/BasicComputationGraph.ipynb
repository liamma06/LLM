This example shows a simple forward pass:

Data goes in -> Some Math -> Prediction

Only for 1 layer and 1 output for a score between 0-1 compared to the true result 0 or 1. 

```python
import torch.nn.functional as F  # grab tool with simplier name 
import torch

y = torch.tensor([1.0]) #target/correct value
x1 = torch.tensor([1.1]) #data fed into model
w1 = torch.tensor([2.2]) #weight of model, the number adjusts through gradient descent
b = torch.tensor([0.0]) #y-int

z = x1 * w1 + b
a = torch.sigmoid(z)#squash value between 0 and 1, good for binary classification
loss = F.binary_cross_entropy(a, y)
print(loss)
print (a)
```

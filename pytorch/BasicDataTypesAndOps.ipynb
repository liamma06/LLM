## Basic Data types and converting to 32 BIT

```python
import torch 

tensor0d = torch.tensor(1)
tensor1d = torch.tensor([1, 2, 3])
tensor2d = torch.tensor([[1, 2], [3,4]])
tensor3d = torch.tensor([[[1, 2], [3,4]], [[5,6], [7,8]]])

```

```python
#better to convert to 32 bit int to save memory
print(tensor1d.dtype)
```

```python
floatvec = tensor1d.to(torch.float32)
print(floatvec.dtype)
print(floatvec)

#tensor([1., 2., 3.]) make it a float 
```

## Common Operations 

```python
import torch 

tensor2d = torch.tensor([[1,2,3], [4,5,6]])

print(tensor2d)
print(tensor2d.shape) #2 rows, 3 columns. 
```

```python
#able to directly reshape
tensor2d_reshaped = tensor2d.reshape(3, 2)
print(tensor2d_reshaped)
```

```python
#also transpose (flipping across diagonals)
tensor2d_transposed = tensor2d.T
print(tensor2d_transposed)
```

```python
#multiply matrices 
tensor2d_multiplied = tensor2d.matmul(tensor2d_transposed)
print(tensor2d_multiplied)

#also can do with @ 
tensor2d_multiplied_2 = tensor2d @ tensor2d_transposed
print(tensor2d_multiplied_2)

```
